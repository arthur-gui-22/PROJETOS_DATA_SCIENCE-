```
texto_completo = " ".join(df["mensagem"])

print(texto_completo)
```

```
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')

tokens = word_tokenize(
    texto_completo.lower(),
    language="portuguese"
)

print(tokens)

print("Quantidade de tokens:")
print(len(tokens))
```

```
from nltk.probability import FreqDist

frequencia_original = FreqDist(tokens)

print("Palavras/tokens mais frequentes:")
print(frequencia_original.most_common(15))
```

```
from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("portuguese"))

palavras_limpas = []

for palavra in tokens:
    if palavra.isalpha():
        if palavra not in stop_words:
            palavras_limpas.append(palavra)

print(palavras_limpas)
```

```
print("Tokens antes da limpeza:")
print(len(tokens))

print("\nPalavras depois da limpeza:")
print(len(palavras_limpas))

print("\nQuantidade removida:")
print(len(tokens) - len(palavras_limpas))
```

```
len(tokens) - len(palavras_limpas)
```

```
 frequencia = FreqDist(palavras_limpas)

print("10 palavras mais frequentes depois da limpeza:")
print(frequencia.most_common(10))
```

```
resultado = pd.DataFrame(
    frequencia.most_common(10),
    columns=["Palavra", "Frequencia"]
)

display(resultado)
```

```
resultado.plot(
    x="Palavra",
    y="Frequencia",
    kind="bar",
    legend=False,
    figsize=(10,5)
)

plt.title("10 palavras mais frequentes nos comentários")
plt.xlabel("Palavras")
plt.ylabel("Frequência")
plt.xticks(rotation=45)
plt.show()
```

```
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report
```

```
X = df["mensagem"]
y = df["categoria"]
```

```
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Mensagens para treinamento:", len(X_treino))
print("Mensagens para teste:", len(X_teste))
```

```
vectorizador = TfidfVectorizer(
    lowercase=True,
    stop_words=list(stop_words),
    ngram_range=(1, 2)
)

X_treino_tfidf = vectorizador.fit_transform(X_treino)
X_teste_tfidf = vectorizador.transform(X_teste)

print("Formato dos dados de treinamento:", X_treino_tfidf.shape)
print("Formato dos dados de teste:", X_teste_tfidf.shape)
```

```
modelo = LogisticRegression(
    max_iter=1000,
    random_state=42
)

modelo.fit(X_treino_tfidf, y_treino)

print("Modelo treinado com sucesso!")
```

```
y_pred = modelo.predict(X_teste_tfidf)

print("Categorias reais:")
print(y_teste.values)

print("\nCategorias previstas:")
print(y_pred)
```

```
acuracia = accuracy_score(y_teste, y_pred)
f1 = f1_score(y_teste, y_pred, average="macro")

print("Acurácia:", round(acuracia, 4))
print("F1-score:", round(f1, 4))
```

```
print(classification_report(y_teste, y_pred))
```

```
categorias = sorted(df["categoria"].unique())

matriz = confusion_matrix(
    y_teste,
    y_pred,
    labels=categorias
)

matriz_df = pd.DataFrame(
    matriz,
    index=categorias,
    columns=categorias
)

display(matriz_df)
```

```
mensagens_surpresa = [
    "Meu cartão foi cobrado, mas o pedido nem foi confirmado.",
    "A encomenda está há dez dias parada.",
    "Comprei o tamanho errado, posso substituir?",
    "O aplicativo não aceita minha senha.",
    "Parabéns, chegou tudo certinho."
]

surpresa_tfidf = vectorizador.transform(mensagens_surpresa)

previsoes = modelo.predict(surpresa_tfidf)
probabilidades = modelo.predict_proba(surpresa_tfidf)

for mensagem, previsao, probabilidade in zip(
    mensagens_surpresa,
    previsoes,
    probabilidades
):
    confianca = max(probabilidade)

    print("Mensagem:", mensagem)
    print("Categoria prevista:", previsao)
    print("Confiança:", round(confianca * 100, 2), "%")
    print("-" * 60)
```

```
mensagem_ambigua = [
    "Meu produto chegou atrasado e quebrado."
]

ambigua_tfidf = vectorizador.transform(mensagem_ambigua)

previsao_ambigua = modelo.predict(ambigua_tfidf)
probabilidade_ambigua = modelo.predict_proba(ambigua_tfidf)

print("Mensagem:", mensagem_ambigua[0])
print("Categoria prevista:", previsao_ambigua[0])
print(
    "Confiança:",
    round(max(probabilidade_ambigua[0]) * 100, 2),
    "%"
)
```
